```python
url = "https://raw.githubusercontent.com/plotly/datasets/master/supermarket_Sales.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Information:")
df.info()
```

```python
df.head()
```

```python
# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

print("\nDate column converted successfully!")

# Check data types again
print(df.dtypes)
```

```python
# Basic statistical summary
df.describe()
```

```python
df.describe()
```

```python
# Product line-wise total sales
product_sales = df.groupby("Product line")["Total"].sum().sort_values(ascending=False)

print(product_sales)
```

```python
product_sales = df.groupby("Product line")["Total"].sum().sort_values(ascending=False)

print(product_sales)
```

```python
plt.figure(figsize=(10,5))

product_sales.plot(kind="bar")

plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
city_sales = df.groupby("City")["Total"].sum().sort_values(ascending=False)

print(city_sales)
```

```python
plt.figure(figsize=(8,5))

city_sales.plot(kind="bar")

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total"].sum()

print(monthly_sales)
```

```python
plt.figure(figsize=(8,5))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
total_revenue = df["Total"].sum()
average_order_value = df["Total"].mean()

print("Total Revenue:", round(total_revenue, 2))
print("Average Order Value:", round(average_order_value, 2))
```

# End-to-End Sales Data Analysis

## Project Overview
This project focuses on analyzing supermarket sales data to understand sales performance, product categories, customer behavior, and monthly sales trends.

The analysis includes data cleaning, exploratory data analysis (EDA), KPI calculation, data visualization, and actionable business insights.

## 1. Dataset and Data Cleaning

The dataset contains 1,000 supermarket sales records with 17 columns, including product line, city, customer type, gender, quantity, total sales, payment method, and date.

### Data Cleaning Performed
- Checked for missing values.
- Checked for duplicate rows.
- Converted the Date column into datetime format.
- Verified the data types of all columns.

### Data Quality Result
- Missing values: 0
- Duplicate rows: 0

The dataset was clean and ready for further analysis.

## 2. Exploratory Data Analysis (EDA)

Exploratory Data Analysis was performed to understand sales patterns and identify important trends in the dataset.

### Product Line Analysis
Food and beverages recorded the highest total sales of 56,144.84, while Health and beauty recorded the lowest total sales of 49,193.74.

### City-wise Analysis
Naypyitaw recorded the highest total sales of 110,568.71, followed by Yangon and Mandalay.

### Monthly Sales Analysis
January 2019 recorded the highest monthly sales of 116,291.87. Sales decreased in February to 97,219.37 and increased again in March to 109,455.51.

### Customer Analysis
Member customers generated total sales of 164,223.44, slightly higher than Normal customers at 158,743.31.

Female customers generated total sales of 167,882.93, while Male customers generated 155,083.82.

## 3. Key Performance Indicators (KPIs)

The following KPIs were calculated to understand the overall sales performance:

- Total Revenue: 322,966.75
- Average Order Value: 322.97
- Total Orders/Records: 1,000

These KPIs provide a quick overview of the overall sales performance of the dataset.

## 4. Data Visualizations

The following visualizations were created to understand the sales performance more clearly.

### 4.1 Total Sales by Product Line

This chart compares total sales across different product categories. Food and beverages recorded the highest sales, while Health and beauty recorded the lowest sales.

### 4.2 Total Sales by City

This chart shows the sales performance of the three cities. Naypyitaw recorded the highest total sales among the cities.

### 4.3 Monthly Sales Trend

This line chart shows the change in sales from January to March 2019. January recorded the highest sales, followed by a decrease in February and a recovery in March.


## 5. Key Insights and Recommendations

### Insight 1: Product Line Performance

Food and beverages generated the highest total sales of 56,144.84.

**Recommendation:** The business can maintain sufficient stock for high-performing products and study their successful sales patterns to improve other product categories.

### Insight 2: City-wise Sales Performance

Naypyitaw recorded the highest total sales of 110,568.71 among the three cities.

**Recommendation:** The business can study the sales strategies and customer patterns in Naypyitaw and apply suitable strategies to other cities.

### Insight 3: Monthly Sales Trend

January recorded the highest monthly sales of 116,291.87, while February recorded the lowest sales of 97,219.37. Sales increased again in March.

**Recommendation:** The business can use targeted promotions and better inventory planning during lower-sales periods to improve monthly performance.
